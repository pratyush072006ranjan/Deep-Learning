```python
quotes=df['quote']
```

```python
quotes=quotes.str.lower()
```

```python
import string
translator=str.maketrans('','',string.punctuation)
```

```python
quotes=quotes.apply(lambda x: x.translate(translator))
```

```python
quotes
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocab_size=10000

tokenizer=Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index=tokenizer.word_index
print(len(word_index))
```

```python
list(word_index.items())[:10]
```

```python
sequence=tokenizer.texts_to_sequences(quotes)
```

```python
quotes[0]
```

```python
sequence[0]
```

```python
X=[]
y=[]

for seq in sequence:
    for i in range(1,len(seq)):
        input_seq=seq[:i]
        output_seq=seq[i]
        X.append(input_seq)
        y.append(output_seq)
```

```python
X
```

```python
len(X)
```

```python
len(y)
```

```python
max_len=max(len(x) for x in X)
print(max_len)

max_index = next(i for i, x in enumerate(X) if len(x) == max_len)

print(max_index)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
X_padded=pad_sequences(X,maxlen=max_len,padding='pre')
```

```python
X_padded
```

```python
y=np.array(y)
```

```python
X_padded.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot=to_categorical(y,num_classes=vocab_size)
```

```python
y.shape
```

```python
y_one_hot.shape
```

### RNN Model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,LSTM,Dense
```

```python
embedding_dim=50
rnn_units=128
```

```python
rnn_model=Sequential()
```

```python
rnn_model.add(Embedding(input_dim=vocab_size,output_dim=embedding_dim,input_length=max_len))

rnn_model.add(
    SimpleRNN(units=rnn_units)
)
rnn_model.add(Dense(units=vocab_size,activation='softmax'))
```

```python
rnn_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
rnn_model.summary()
```

### LSTM Model

```python
lstm_model=Sequential()
```

```python
lstm_model.add(
    Embedding(input_dim=vocab_size,output_dim=embedding_dim,input_length=max_len)
)

lstm_model.add(
    LSTM(units=rnn_units)
)

lstm_model.add(Dense(
    units=vocab_size,activation='softmax'
))
```

```python
lstm_model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
lstm_model.summary()
```

```python
epochs=10
batch_size=128
```

```python
history_rnn=rnn_model.fit(
    X_padded,y_one_hot,
    epochs=epochs,
    batch_size=batch_size,
    validation_split=0.1

)
```
